# Forecast Visualizations

Visualize aligned predictions and the recorded benchmark metrics.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DATA_DIR = PROJECT_ROOT / "data" / "processed"
MODEL_DIR = PROJECT_ROOT / "models"

files = {
    "Baseline": ("baseline_predictions.csv", "Prediction"),
    "ARIMA": ("arima_predictions.csv", "ARIMA_Prediction"),
    "Auto ARIMA": ("auto_arima_predictions.csv", "Auto_ARIMA_Prediction"),
    "LSTM": ("lstm_predictions.csv", "LSTM_Prediction"),
    "Prophet": ("prophet_predictions.csv", "Prophet_Prediction"),
}
```

```python
predictions = {}
for model_name, (filename, prediction_column) in files.items():
    frame = pd.read_csv(DATA_DIR / filename, parse_dates=["Date"])
    predictions[model_name] = (frame, prediction_column)
```

```python
fig, axes = plt.subplots(len(predictions), 1, figsize=(14, 18), sharex=True)
for axis, (model_name, (frame, prediction_column)) in zip(axes, predictions.items()):
    axis.plot(frame["Date"], frame["Actual"], label="Actual")
    axis.plot(frame["Date"], frame[prediction_column], label=model_name)
    axis.set_title(model_name)
    axis.set_ylabel("INR")
    axis.legend()
axes[-1].set_xlabel("Date")
fig.tight_layout()
plt.show()
```

```python
comparison = pd.read_csv(DATA_DIR / "model_comparison.csv")
fig, ax = plt.subplots(figsize=(10, 5))
ax.bar(comparison["Model"], comparison["RMSE"])
ax.set(title="RMSE on Shared Evaluation Window", xlabel="Model", ylabel="RMSE")
ax.tick_params(axis="x", rotation=20)
plt.show()
```

```python
lstm_frame, lstm_column = predictions["LSTM"]
fig, ax = plt.subplots(figsize=(14, 7))
ax.plot(lstm_frame["Date"], lstm_frame["Actual"], label="Actual")
ax.plot(lstm_frame["Date"], lstm_frame[lstm_column], label="LSTM")
ax.set(title="LSTM Forecast vs Actual", xlabel="Date", ylabel="Closing Price (INR)")
ax.legend()
fig.tight_layout()
fig.savefig(MODEL_DIR / "lstm_forecast_plot.png", dpi=120)
plt.show()
```
