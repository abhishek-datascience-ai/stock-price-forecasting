```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("seaborn-v0_8")
```

```python
baseline = pd.read_csv("../data/processed/baseline_predictions.csv")
arima = pd.read_csv("../data/processed/arima_predictions.csv")
auto_arima = pd.read_csv("../data/processed/auto_arima_predictions.csv")
lstm = pd.read_csv("../data/processed/lstm_predictions.csv")
prophet = pd.read_csv("../data/processed/prophet_predictions.csv")
```

```python
plt.figure(figsize=(12,6))

plt.plot(baseline["Actual"], label="Actual")
plt.plot(baseline["Prediction"], label="Baseline")

plt.title("Baseline Forecast vs Actual")
plt.legend()

plt.show()
```

```python
plt.figure(figsize=(12,6))

plt.plot(arima["Actual"], label="Actual")
plt.plot(arima["ARIMA_Prediction"], label="ARIMA")

plt.title("ARIMA Forecast vs Actual")
plt.legend()

plt.show()
```

```python
plt.figure(figsize=(12,6))

plt.plot(auto_arima["Actual"], label="Actual")
plt.plot(auto_arima["Auto_ARIMA_Prediction"], label="Auto ARIMA")

plt.title("Auto ARIMA Forecast vs Actual")
plt.legend()

plt.show()
```

```python
plt.figure(figsize=(12,6))

plt.plot(lstm["Actual"], label="Actual")
plt.plot(lstm["LSTM_Prediction"], label="LSTM")

plt.title("LSTM Forecast vs Actual")
plt.legend()

plt.show()
```

```python
plt.figure(figsize=(12,6))

plt.plot(prophet["Actual"], label="Actual")
plt.plot(prophet["Prophet_Prediction"], label="Prophet")

plt.title("Prophet Forecast vs Actual")
plt.legend()

plt.show()
```

```python
plt.figure(figsize=(14,7))

plt.plot(baseline["Actual"], label="Actual")

plt.plot(baseline["Prediction"], label="Baseline")
plt.plot(arima["ARIMA_Prediction"], label="ARIMA")
plt.plot(auto_arima["Auto_ARIMA_Prediction"], label="Auto ARIMA")
plt.plot(lstm["LSTM_Prediction"], label="LSTM")
plt.plot(prophet["Prophet_Prediction"], label="Prophet")

plt.title("All Model Predictions Comparison")

plt.legend()

plt.show()
```

#### Actual - Prediction

```python
lstm["Residual"] = lstm["Actual"] - lstm["LSTM_Prediction"]
```

```python
plt.figure(figsize=(10,6))

sns.histplot(lstm["Residual"], bins=50, kde=True)

plt.title("LSTM Residual Distribution")

plt.show()
```
